# Generation of Gaussian Process based availability matrices

In this notebook we generate various availability matrices by means of Gaussian Processes, with and without a fine-tuning phase at the end.

At the moment this code accounts for time correlation, with homogeneous or hetergogeneous participation frequencies. It does not account for spatial correlation.

**Import libraries:**

```python
from av_mat_generation.GP_based.gp import GP
import numpy as np

MAIN_FOLDER = 'availability_matrices/av-mat-6'
```

## Generation of availablity matrices

```python
freq_seq = np.array([0.25, 0.25, 0.50, 0.50, 0.50, 0.75, 0.75])
gp = GP(freq=freq_seq, out_folder=MAIN_FOLDER, n_rounds=50, k=3)
gp.create_all_av_mat()

freq_seq = 0.35
gp = GP(freq=freq_seq, out_folder=MAIN_FOLDER, n_rounds=50, k=3)
gp.create_all_av_mat()

freq_seq = 0.5
gp = GP(freq=freq_seq, out_folder=MAIN_FOLDER, n_rounds=50, k=3)
gp.create_all_av_mat()
```
